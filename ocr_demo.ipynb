# OCR Recognition NN - Demo Notebook

Demonstration der DeepSeek OCR Pipeline.

```python
from dsocr import Pipeline
from PIL import Image
import numpy as np

# Demo: Erstelle ein Testbild
img = Image.fromarray(np.zeros((100, 100, 3), dtype=np.uint8))
print(f"Bildgroesse: {img.size}")
```

```python
# Pipeline mit Demo-Engine
pipeline = Pipeline(backend="demo")
result = pipeline.run([img])
print(f"Seiten: {len(result.pages)}")
print(f"Erkannte Regionen: {len(result.pages[0].regions)}")
```
